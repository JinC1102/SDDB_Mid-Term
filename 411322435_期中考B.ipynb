````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
conn = sqlite3.connect("cafe.db")
cursor = conn.cursor()

# 重建資料表
cursor.execute("DROP TABLE IF EXISTS menu")
cursor.execute("DROP TABLE IF EXISTS orders")

cursor.execute("""
CREATE TABLE menu (
    name TEXT PRIMARY KEY,
    price INTEGER NOT NULL,
    stock INTEGER NOT NULL
)
""")

cursor.execute("""
CREATE TABLE orders (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    student_name TEXT NOT NULL,
    item_name TEXT NOT NULL,
    quantity INTEGER NOT NULL,
    status TEXT NOT NULL
)
""")

# 寫入初始資料
initial_menu = [
    ("美式咖啡", 50, 5),
    ("拿鐵", 65, 4),
    ("抹茶歐蕾", 70, 3)
]

cursor.executemany("INSERT INTO menu (name, price, stock) VALUES (?, ?, ?)", initial_menu)
conn.commit()
conn.close()
print("✅ 資料庫初始化完成，初始菜單已載入！")
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
def show_tables():
    conn = sqlite3.connect("cafe.db")
    print("=== 菜單資料表 (menu) ===")
    df_menu = pd.read_sql_query("SELECT * FROM menu", conn)
    display(df_menu)

    print("\n=== 訂單資料表 (orders) ===")
    df_orders = pd.read_sql_query("SELECT * FROM orders", conn)
    display(df_orders)
    conn.close()

show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 品項, 數量, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
ai_cache = {}

def parse_user_intent(user_text):
    if user_text in ai_cache:
        return ai_cache[user_text]

    system_prompt = (
        "你是一個校園咖啡廳預約系統助手。請解析使用者的句子，並嚴格只回覆一行格式：\n"
        "動作,品項,數量\n\n"
        "規則：\n"
        "1. 動作只能是【訂購】、【取消】、【查詢】、【統計】、【未知】之一。\n"
        "2. 品項只能是【美式咖啡】、【拿鐵】、【抹茶歐蕾】、【無】之一。\n"
        "3. 數量為正整數（如 1, 2, 3），若句中用文字如『兩杯』請轉為數字 2；若未提及數量預設為 1。\n"
        "4. 若使用者詢問銷售狀況、營業額、賣得怎樣等，動作回覆【統計】，品項為【無】。\n"
        "5. 無法判斷時，動作回覆【未知】，品項為【無】。\n"
        "絕對不要加入任何額外說明文字。"
    )

    raw_reply = ""
    max_retries = 3
    for attempt in range(max_retries):
        try:
            response = client.chat.completions.create(
                model=MODEL,
                messages=[
                    {"role": "system", "content": system_prompt},
                    {"role": "user", "content": user_text}
                ],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )
            raw_reply = response.choices[0].message.content.strip()
            time.sleep(2)
            break
        except Exception as e:
            if "429" in str(e) and attempt < max_retries - 1:
                print(f"⚠️ 觸發 Rate Limit，等待 30 秒後重試（第 {attempt + 1} 次）...")
                time.sleep(30)
            else:
                raw_reply = f"Error: {e}"
                break

    # 後處理校驗（不完全相信 AI）
    action = "未知"
    item = "無"
    quantity = 1

    valid_actions = ["訂購", "取消", "查詢", "統計"]
    valid_items = ["美式咖啡", "拿鐵", "抹茶歐蕾"]

    for act in valid_actions:
        if act in raw_reply:
            action = act
            break

    for it in valid_items:
        if it in raw_reply:
            item = it
            break

    # 解析數量數字
    match = re.search(r'\d+', raw_reply)
    if match:
        quantity = int(match.group())
    else:
        quantity = 1

    result = (action, item, quantity, raw_reply)
    ai_cache[user_text] = result
    return result
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
test_cases = [
    ("我想訂一杯拿鐵", "訂購", "拿鐵", 1),
    ("我要兩杯抹茶歐蕾", "訂購", "抹茶歐蕾", 2),
    ("幫我取消美式咖啡", "取消", "美式咖啡", 1),
    ("請問拿鐵還有幾杯", "查詢", "拿鐵", 1),
    ("今天各品項賣得怎樣", "統計", "無", 1),
    ("訂三杯美式咖啡", "訂購", "美式咖啡", 3),
    ("我想買5杯拿鐵", "訂購", "拿鐵", 5),
    ("隨便給我來點吃的", "未知", "無", 1),
    ("我要一杯珍珠奶茶", "未知", "無", 1)
]

passed_count = 0
print(f"{'#':<3} | {'使用者輸入':<18} | {'預期結果':<15} | {'AI原始回覆':<15} | {'程式判定結果':<15} | {'判定'}")
print("-" * 85)

for idx, (text, exp_act, exp_item, exp_qty) in enumerate(test_cases, 1):
    act, item, qty, raw = parse_user_intent(text)

    # 檢查判定 logic
    match_act = (exp_act == "*") or (act == exp_act)
    match_item = (exp_item == "*") or (item == exp_item)
    match_qty = (exp_qty == "*") or (qty == exp_qty)

    if match_act and match_item and match_qty:
        status = "✅ 通過"
        passed_count += 1
    else:
        status = "❌ 失敗"

    exp_str = f"{exp_act},{exp_item},{exp_qty}"
    got_str = f"{act},{item},{qty}"
    print(f"{idx:<3} | {text:<18} | {exp_str:<15} | {raw:<15} | {got_str:<15} | {status}")

print("-" * 85)
print(f"📊 測試總結：通過 {passed_count} / {len(test_cases)} 句測試。")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
def process_order(name, user_text):
    action, item, qty, raw_reply = parse_user_intent(user_text)

    if action == "未知":
        return "❌ 無法識別需求。目前提供品項：美式咖啡、拿鐵、抹茶歐蕾。"

    conn = sqlite3.connect("cafe.db")
    cursor = conn.cursor()

    try:
        if action == "統計":
            # 使用 JOIN 與 GROUP BY 進行統計
            query = """
            SELECT
                m.name,
                m.price,
                m.stock,
                COALESCE(SUM(CASE WHEN o.status = '已成立' THEN o.quantity ELSE 0 END), 0) AS sold_qty,
                (COALESCE(SUM(CASE WHEN o.status = '已成立' THEN o.quantity ELSE 0 END), 0) * m.price) AS total_revenue
            FROM menu m
            LEFT JOIN orders o ON m.name = o.item_name
            GROUP BY m.name, m.price, m.stock
            """
            cursor.execute(query)
            rows = cursor.fetchall()

            res = "📊 【今日銷售統計報告】\n"
            res += f"{'品項':<8} | {'單價':<5} | {'已售':<5} | {'剩餘':<5} | {'營業額':<6}\n"
            res += "-" * 40 + "\n"
            total_sum = 0
            for r in rows:
                p_name, price, stock, sold, rev = r
                res += f"{p_name:<8} | ${price:<4} | {sold:<5}杯 | {stock:<5}杯 | ${rev:<6}\n"
                total_sum += rev
            res += "-" * 40 + "\n"
            res += f"💰 總營業額：{total_sum} 元"
            return res

        elif action == "查詢":
            if item == "無":
                return "請說明欲查詢的品項名稱。"
            cursor.execute("SELECT stock FROM menu WHERE name = ?", (item,))
            row = cursor.fetchone()
            if row is None:
                return f"找不到品項：{item}"
            return f"ℹ️ 【{item}】目前剩餘庫存為 {row[0]} 杯。"

        elif action == "訂購":
            if not name:
                return "❌ 訂購需要提供姓名。"
            if item == "無":
                return "❌ 無法識別您欲購買的品項。"
            if qty < 1 or qty > 3:
                return "❌ 訂購失敗：單次預約數量需在 1 ~ 3 杯之間。"

            # 檢查個人上限 (每人當日最多 3 杯，已取消不計)
            cursor.execute(
                "SELECT SUM(quantity) FROM orders WHERE student_name = ? AND status = '已成立'",
                (name,)
            )
            user_bought_row = cursor.fetchone()
            current_user_qty = user_bought_row[0] if (user_bought_row and user_bought_row[0]) else 0

            if current_user_qty + qty > 3:
                return f"❌ 訂購失敗：{name} 今日已訂購 {current_user_qty} 杯，每人每日最多預約 3 杯。"

            # 檢查庫存與單價
            cursor.execute("SELECT price, stock FROM menu WHERE name = ?", (item,))
            menu_row = cursor.fetchone()
            if menu_row is None:
                return f"❌ 找不到品項：{item}"

            price, stock = menu_row
            if stock < qty:
                return f"❌ 訂購失敗：{item} 今日剩餘庫存不足（剩餘 {stock} 杯）。"

            # 扣減庫存與寫入訂單
            cursor.execute("UPDATE menu SET stock = stock - ? WHERE name = ?", (qty, item))
            cursor.execute(
                "INSERT INTO orders (student_name, item_name, quantity, status) VALUES (?, ?, ?, '已成立')",
                (name, item, qty)
            )
            conn.commit()
            total_price = price * qty
            return f"🎉 預約成功！{name} 預約 {item} {qty} 杯，共 {total_price} 元。"

        elif action == "取消":
            if not name:
                return "❌ 取消訂單需要提供姓名。"
            if item == "無":
                return "❌ 無法識別您欲取消的品項。"

            # 查詢最新一筆已成立的訂單
            cursor.execute(
                "SELECT id, quantity FROM orders WHERE student_name = ? AND item_name = ? AND status = '已成立' ORDER BY id DESC LIMIT 1",
                (name, item)
            )
            order_row = cursor.fetchone()
            if order_row is None:
                return f"❌ 找不到 {name} 預約 【{item}】 的有效訂單。"

            order_id, cancel_qty = order_row
            # 修改訂單狀態並退回庫存
            cursor.execute("UPDATE orders SET status = '已取消' WHERE id = ?", (order_id,))
            cursor.execute("UPDATE menu SET stock = stock + ? WHERE name = ?", (cancel_qty, item))
            conn.commit()
            return f"✅ 已成功取消 {name} 的 【{item}】 {cancel_qty} 杯預約，庫存已釋出。"

    finally:
        conn.close()
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
flow_test_cases = [
    ("小明", "我想訂 2 杯拿鐵"),
    ("小明", "請問拿鐵還有幾杯"),
    ("小明", "再訂 2 杯拿鐵"),
    ("小華", "我要訂 3 杯美式咖啡"),
    ("小華", "我要訂 3 杯抹茶歐蕾"),
    ("小華", "再訂 1 杯抹茶歐蕾"),
    ("小明", "取消拿鐵"),
    ("小美", "我要取消美式咖啡"),
    ("店長", "今天各品項賣得怎樣")
]

print("🚀 開始執行流程測試...\n")
for idx, (name, sentence) in enumerate(flow_test_cases, 1):
    print(f"--- [步驟 {idx}] 姓名：{name} ｜ 句子：『{sentence}』 ---")
    response_msg = process_order(name, sentence)
    print(f"系統回應：\n{response_msg}\n")

print("=" * 50)
print("📋 流程測試執行完畢，最終資料庫狀態：")
print("=" * 50)
show_tables()
````
